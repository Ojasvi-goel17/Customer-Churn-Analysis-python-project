```python
# converting datatype and replace value with null
df["TotalCharges"]=df["TotalCharges"].replace(" ","0")
df["TotalCharges"]=df["TotalCharges"].astype("float")
```

```python
# convert 0 and 1 value to yes and no t make it easy to understand
def conv(value):
    if value==1:
        return"yes"
    else:
        return"no"

df["SeniorCitizen"]=df["SeniorCitizen"].apply(conv)
```

```python
# to show count of customer using the servive or who have left the service
# to show the number above the bar
import seaborn as sns
import matplotlib.pyplot as plt
ax = sns.countplot(x =df['Churn'],data =df)
ax.bar_label(ax.containers[0])
plt.title("count of customeres by churn")


```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(3,4))
gb=df.groupby("Churn").agg({'Churn':"count"})
plt.pie(gb['Churn'],labels=gb.index,autopct="%1.2f%%")
plt.title("Percentage of Churned Costomer",fontsize=10)
plt.show
```

```python
plt.figure(figsize=(3,4))
sns.countplot(x="SeniorCitizen",data=df,hue="Churn")
plt.title("Churn by SeniorCitizen")
```

```python

# Step 1: Group and calculate percentages
grouped = df.groupby(['SeniorCitizen', 'Churn']).size().unstack(fill_value=0)
percentages = grouped.div(grouped.sum(axis=1), axis=0) * 100

# Step 2: Plot stacked bar chart
fig, ax = plt.subplots(figsize=(3, 3))

bottom = [0] * len(percentages)
for churn_status in percentages.columns:
    values = percentages[churn_status]
    ax.bar(percentages.index, values, bottom=bottom, label=churn_status)

    # Add percentage labels
    for i, (val, btm) in enumerate(zip(values, bottom)):
        ax.text(i, btm + val / 2, f'{val:.1f}%', ha='center', va='center',
                fontsize=9, color='white')
        
    bottom = [btm + val for btm, val in zip(bottom, values)]

# Step 3: Customize plot
ax.set_ylabel('Percentage of Customers')
ax.set_title('Churn by Senior Citizen Status')
ax.set_xticks([0, 1])
ax.set_xticklabels(['Not Senior', 'Senior'])
ax.legend(title='Churn')
plt.tight_layout()
plt.show()


```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# List of columns to plot
cols = ['PhoneService', 'MultipleLines', 'InternetService',
        'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
        'TechSupport', 'StreamingTV', 'StreamingMovies']

# Set up the subplot grid
rows = 3
cols_per_row = 3
fig, axes = plt.subplots(rows, cols_per_row, figsize=(18, 12))
axes = axes.flatten()  # Flatten to easily iterate

# Create countplots
for i, col in enumerate(cols):
    sns.countplot(x=col, data=df, ax=axes[i],hue="Churn")
    axes[i].set_title(f'Countplot of {col}')
    axes[i].tick_params(axis='x', rotation=45)

# Remove any unused subplots
for j in range(len(cols), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

#The charts reveal that customers lacking services like OnlineSecurity, TechSupport, and DeviceProtection tend to churn more frequently. Fiber optic internet users also show higher churn rates compared to DSL or no internet service. In contrast, customers with PhoneService and MultipleLines generally exhibit lower churn. These insights highlight which service gaps may be contributing to customer attrition.
